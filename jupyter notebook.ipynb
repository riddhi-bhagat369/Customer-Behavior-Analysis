```python
import pandas as pd
df = pd.read_csv("customer_shopping_behavior.csv")
```

```python
df.head(10)
```

```python
df.info()
```

```python
#df.describe() - summary statistics of numerical columns
df.describe(include='all')
```

```python
df.isnull().sum()
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
#snake casing
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
```

```python
df.columns
```

```python
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#create a column of age_group
label = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = label)
```

```python
df[['age','age_group']].head(10)
```

```python
#purchase_frequency_days
df['frequency_of_purchases'].unique()
```

```python
frequency_mapping = {'Fortnightly' : 14, 'Weekly' : 7, 'Annually' : 365, 'Quarterly' : 90, 'Bi-Weekly' : 14, 'Monthly' : 30, 'Every 3 Months' : 90}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_purchases', 'purchase_frequency_days']].head(10)
```

```python
df[['discount_applied', 'promo_code_used']].head()
```
